# Adivinanza

```python
import random
# El programa genera un número aleatorio entre 1 y 100 y le pide al usuario que adivine el número. El programa le dará pistas al usuario si su suposición es demasiado alta o demasiado baja, y continuará hasta que el usuario adivine correctamente. 
numero_secreto = random.randint(1,100)
intentos = 0 
print("¡Bienvenido al juego de adivinanza de números!") 
print("He elegido un número entre el 1 y el 100. ¿Puedes adivinar cuál es?\n")
while True:
    intento = input("Introduce tu suposición: ")
    intentos += 1
    try:
        intento = int(intento)
        if intento < numero_secreto:
            print("Demasiado bajo. Inténtalo de nuevo.")
        elif intento > numero_secreto:
            print("Demasiado alto. Inténtalo de nuevo.")
        else:
            print(f"¡Felicidades! Has adivinado el número en {intentos} intentos.")
            break
    except ValueError:
        print("Por favor, introduce un número válido.")
```
